```python
# prompt: use python bigquery client to run a SQL query. The query must have a parameter 'nct_number'

from google.cloud import bigquery

def query_nct(nct_number):
  client = bigquery.Client()
  query = """
  SELECT * FROM (
  SELECT
  *
  FROM
    `meridian-dev-455515`.`clinical_trial`.`ClinicalTrialMasterData` AS t
  QUALIFY
    ROW_NUMBER() OVER (PARTITION BY t.PostingID ORDER BY 1) = 1
  )
  WHERE NCT_Number = @nct_number
  """
  job_config = bigquery.QueryJobConfig(
      query_parameters=[
          bigquery.ScalarQueryParameter("nct_number", "STRING", nct_number),
      ]
  )
  query_job = client.query(query, job_config=job_config)
  results = query_job.result()
  query_output = ""
  for row in results:
    query_output += str(row)
  df = query_job.result().to_dataframe()
  #print(f"\n✅ Returned {len(df)} rows\n")
  return df
```

```python
from google.cloud import storage

def check_file_exists_in_gcs(nct_number_input, posting_id_input):
  """
  Checks if a file with a specific NCT_Number and PostingID exists in GCS.

  Args:
      nct_number_input (str): The NCT_Number to search for.
      posting_id_input (str): The PostingID to search for.

  Returns:
      bool: True if the file exists, False otherwise.
  """
  storage_client = storage.Client()
  bucket_name = 'cssr_reports_trials'
  folder_name = 'april2_generation'
  bucket = storage_client.bucket(bucket_name)
  blobs = bucket.list_blobs(prefix=folder_name)

  for blob in blobs:
    file_name = blob.name
    # Extract PostingID and NCT_Number from the filename
    try:
      # Get the actual filename without path and extension
      base_name = file_name.split('/')[-1].replace('.pdf', '')
      parts = base_name.split('_')

      if len(parts) == 2:
        posting_id = parts[0]
        nct_number = parts[1]

        # Fix: Convert both to string to avoid type mismatch (int vs str)
        if str(nct_number) == str(nct_number_input) and str(posting_id) == str(posting_id_input):
          return True
    except Exception as e:
      print(f"Error parsing filename {file_name}: {e}")
      continue
  return False
```
